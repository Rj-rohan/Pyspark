# Aggregations in PySpark

```python
from pyspark.sql.functions import *
```

```python
data = [
    (1,"rohan","pune",'IT',10000),
    (2,"tushar","nagpur",'HR',20000),
    (3,"ansh","pune",'Finance',30000),
    (4,"prasad","mumbai",'HR',40000),
    (5,"mohit","nashik",'IT',50000),
    
]

df = spark.createDataFrame(data,['id','name','city','dept','salary'])
df.display()
```

## Mostly it is used with aggregation functions

## GroupBy - Used to group rows based on a column.


### 1.Count

```python
df.groupBy("city").count().show()
```

### 2. sum

```python
df.groupBy("dept").sum("salary").show()
```

### 3. avg

```python
df.groupBy("dept").avg("salary").show()
```

### 4. max

```python
df.groupBy("dept").max("salary").show()
```

### 5. min

```python
df.groupBy("dept").min("salary").show()
```

### agg() - Used when multiple aggregations are needed together.

```python
df.groupBy("dept").agg(
    sum(col("salary")).alias("total_salary"),
    avg(col("salary")).alias("avg_salary"),
    max(col("salary")).alias("max_salary"),
    min(col("salary")).alias("min_salary")
).show()
```

### Having-like Condition In SQL:

```python
# Means filtering records after condition
df.groupBy('dept').count().filter(col('count') > 1).show()
```
