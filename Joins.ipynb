# JOINS In PySpark

```python
from pyspark.sql.functions import *
```

## Create df1

```python
emp_data = [
    (1, "Rohan", "Pune", "IT", 50000),
    (2, "Amit", "Mumbai", "HR", 40000),
    (3, "Priya", "Delhi", "Finance", 60000),
    (4, "Sneha", "Pune", "IT", 70000),
    (5, "Karan", "Nagpur", "Admin", 45000)  
]

df1 = spark.createDataFrame(
    emp_data,
    ["id", "name", "city", "dept", "salary"]
)

df1.display()
```

## Create df2

```python
dept_data = [
    ("IT", "Technology"),
    ("HR", "Human Resource"),
    ("Finance", "Accounts"),
    ("Marketing", "Sales Team")   # Marketing has no employee
]

df2 = spark.createDataFrame(
    dept_data,
    ["dept", "dept_full_name"]
)

df2.display()
```

### Inner Join

```python
# Only matching records come.
df1.join(df2,on="dept",how="inner").display()
```

```python
# trying to use other query with join
df1.join(df2,on="dept",how="inner").select(col('name')).display()
```

### Left Join

```python
df1.join(df2,on="dept",how="left").display()
```

### Right Join

```python
df1.join(df2, on="dept", how="right").display()
```

### Full Outer Join

```python
# All records from both tables
df1.join(df2, on="dept", how="full").display()
```

### Left Semi Join

```python
# Only matching rows from LEFT table
# Only columns from left table come.
df1.join(df2, on="dept", how="left_semi").display()
```

### Left Anti Join

```python
# Only non-matching rows from LEFT table
# Show employees whose department does NOT exist in dept table.
df1.join(df2, on="dept", how="left_anti").display()
```

### Cross Join

```python
# cross join
df1.crossJoin(df2).display()
```
