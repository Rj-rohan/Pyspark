## SQL Functions in PySpark

```python
from pyspark.sql.functions import *
```

```python
data = [
    (1,"rohan","pune",10000),
    (2,"tushar","nagpur",20000),
    (3,"ansh","pune",30000),
    (4,"prasad","mumbai",40000),
    (5,"mohit","nashik",50000),
    (5,"mohit","nashik",50000)
]

df = spark.createDataFrame(data,['id','name','city','salary'])
df.display()
```

```python
# col() - Used to refer to a column properly.
df.select(col("city")).show()
```

```python
# lit() - Used to add a constant (fixed value) column.
df.withColumn("country",lit("India")).show()
```

```python
# when() - Used like IF-ELSE condition.
df.withColumn('salary_status',when(col('salary')>30000,'HIGH').otherwise("LOW")).show()
```

```python
# concat() - Used to join strings.
df.withColumn('full_info',concat(col('name'),col("city"))).show()

# Better version with space:
df.withColumn('full_info',concat(col('name'),lit(' '),col("city"))).show()
```

```python
# SubString() - Used to extract part of a string.
df.withColumn('name',substring(col('name'),0,3)).show()

#The 0 (Start Position) This tells PySpark where to start cutting.
# In PySpark, both 0 and 1 usually refer to the first character of the string.

# The 3 (Length)
# This tells PySpark how many characters to take from the starting point.
```

```python
# lower() - lowercase
df.withColumn(
    "city_lower",
    lower(col("city"))
).show()
```

```python
# upper() - uppercase
df.withColumn(
    "city_lower",
    upper(col("city"))
).show()
```

```python
# trim() - to remove extra space
df.withColumn(
    "name",
    trim(col("name"))
).show()
```

```python
# current_date - gives todays date
df.withColumn(
    "load_date",
    current_date()
).show()
```
