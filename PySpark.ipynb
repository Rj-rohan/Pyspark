## Basics of PySpark


```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import *
```

## List Of Tuples and creating DataFrames

```python
data = [
    (1,"rohan","pune",10000),
    (2,"tushar","nagpur",20000),
    (3,"ansh","pune",30000),
    (4,"prasad","mumbai",40000),
    (5,"mohit","nashik",50000),
    (5,"mohit","nashik",50000)
]

df = spark.createDataFrame(data,['id','name','city','salary'])
df.display()
```

## Dataframe Operations

```python
df.printSchema()
df.columns
df.dtypes
print(type(data))
```

```python
# Select specific columns
df.select('name','salary').show() 
# df.select(col('name')).show() # Other method to write
```

```python
# filter() used to apply condition
df.select(col('name')).filter(df.salary>30000).show()
```

```python
# where() used to apply condition
df.select(col('name')).where(df.salary>30000).show() # There is no diff b/w where and filter
```

```python
# withColumn() is Used to create new column or modify existing column
df1 = df.withColumn('bonus',col('salary')+10000).filter(col('bonus')>50000)
df1.show()
## In PySpark, DataFrames are immutable. When you run df.withColumn(...), you aren't actually adding a column to the variable df; you are creating a new temporary DataFrame.

# df.salary works because salary exists in the original variable.

# df.bonus fails because the variable df still points to the old version of the data that only has id, name, city, and salary.
```

```python
# drop() is used to drop column
df1.drop(col('bonus')).show()
```

```python
# Rename the column
df1.withColumnRenamed('salary','salary_new').show()
```

```python
# distinct() - Remove duplicate rows
df1.distinct().show()

```

```python
# dropDuplicates() - Remove duplicates based on specific columns , Keeps the first record only
df.dropDuplicates(["city"]).show()
```

```python
# orderBy() - Sorting data, default is Ascending
df.orderBy(col("salary")).show()
df.orderBy(col("salary").desc()).show()
```

```python
# limit() - get some rows only
df.limit(4).show()
```
